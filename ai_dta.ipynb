```python
"""
AI Project Failure Analysis — Data Generator
Why AI Projects Fail: A Data-Driven Analysis of Organizational Adoption
"""
import pandas as pd
import numpy as np
import random
```

```python

np.random.seed(42)
random.seed(42)
n = 300

industries     = ['Financial Services','Healthcare','Retail','Manufacturing',
                  'Technology','Government','Education','Energy']
sizes          = ['Small (<200)','Mid-size (200-1000)',
                  'Large (1000-5000)','Enterprise (5000+)']
regions        = ['North America','Europe','Asia-Pacific',
                  'Latin America','Middle East & Africa']
project_types  = ['NLP/Chatbot','Computer Vision','Predictive Analytics',
                  'Recommendation Engine','Process Automation',
                  'Generative AI','Fraud Detection']
failure_reasons= [
    'Poor Data Quality','Lack of Skilled Talent','No Clear Business Use Case',
    'Change Management Issues','Insufficient Budget',
    'Regulatory/Compliance Hurdles','Model Performance Gaps',
    'Siloed Organizational Structure','Vendor/Tech Dependency',
    'Lack of Executive Sponsorship'
]
outcomes = ['Failed','Partial Failure','Ongoing (At Risk)','Succeeded']

data = []
for i in range(n):
    industry    = random.choice(industries)
    size        = random.choice(sizes)
    region      = random.choice(regions)
    proj_type   = random.choice(project_types)
    outcome     = np.random.choice(outcomes, p=[0.35, 0.25, 0.15, 0.25])

    budget_ranges = {
        'Small (<200)'        : (0.1, 0.5),
        'Mid-size (200-1000)' : (0.5, 2.0),
        'Large (1000-5000)'   : (2.0, 10.0),
        'Enterprise (5000+)'  : (5.0, 50.0)
    }
    lo, hi        = budget_ranges[size]
    budget_usd_m  = round(np.random.uniform(lo, hi), 2)
    duration_months = int(max(3, min(48, np.random.normal(18, 6))))

    primary_reason   = random.choice(failure_reasons)
    secondary_reason = random.choice([r for r in failure_reasons if r != primary_reason])

    def score(mu_succ, mu_fail, sd_succ=1.0, sd_fail=1.5):
        mu = mu_succ if outcome == 'Succeeded' else mu_fail
        sd = sd_succ if outcome == 'Succeeded' else sd_fail
        return int(max(1, min(10, np.random.normal(mu, sd))))

    data_readiness       = score(7.5, 4.5)
    leadership_support   = score(7.8, 4.2)
    talent_score         = int(max(1, min(10, np.random.normal(5.5, 2))))
    change_mgmt          = score(7.2, 4.0, 1.2, 1.5)

    year_launched = random.choice([2019, 2020, 2021, 2022, 2023, 2024])

    roi_params = {
        'Succeeded'        : (145, 40),
        'Partial Failure'  : (45, 20),
        'Failed'           : (-30, 20),
        'Ongoing (At Risk)': None
    }
    p = roi_params[outcome]
    roi_pct = round(np.random.normal(*p), 1) if p else None

    has_ai_team   = np.random.choice([True, False],
                        p=[0.72, 0.28] if outcome == 'Succeeded' else [0.45, 0.55])
    used_vendor   = np.random.choice([True, False])
    has_governance= np.random.choice([True, False],
                        p=[0.35, 0.65] if outcome == 'Failed' else [0.60, 0.40])

    data.append({
        'project_id'              : f'AIPROJ-{1000+i}',
        'industry'                : industry,
        'company_size'            : size,
        'region'                  : region,
        'project_type'            : proj_type,
        'outcome'                 : outcome,
        'budget_usd_m'            : budget_usd_m,
        'duration_months'         : duration_months,
        'primary_failure_reason'  : primary_reason,
        'secondary_failure_reason': secondary_reason,
        'data_readiness_score'    : data_readiness,
        'leadership_support_score': leadership_support,
        'talent_score'            : talent_score,
        'change_mgmt_score'       : change_mgmt,
        'year_launched'           : year_launched,
        'roi_pct'                 : roi_pct,
        'has_dedicated_ai_team'   : has_ai_team,
        'used_external_vendor'    : used_vendor,
        'has_data_governance'     : has_governance
    })
```

```python
df = pd.DataFrame(data)
df.to_csv('ai_projects.csv', index=False)
print(f"Generated {len(df)} records.")
print(df['outcome'].value_counts())
```
